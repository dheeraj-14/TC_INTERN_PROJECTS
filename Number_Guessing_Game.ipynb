```python
import random
print("=" * 40)
print("WELCOME TO THE NUMBER GUESSING GAME")
print("=" * 40)
while True:
    number = random.randint(1, 100)
    chances = 7
    print("\nThe computer has selected a number between 1 and 100.")
    print("You have", chances, "chances to guess it correctly.")
    for count in range(1, chances+1):
        print("\nAttempt", count)
        try:
            guess = int(input("Enter your guess: "))
            if guess < 1 or guess > 100:
                print("Please enter a number between 1 and 100.")
                continue
        except ValueError:
            print("Invalid input! Enter numbers only.")
            continue
        if guess < number:
            print("Too Low!")
        elif guess > number:
            print("Too High!")
        else:
            print("\nCorrect!")
            print("You guessed the number in", count, "attempts.")
            break
        if count != chances:
            print("Attempts Left:", chances - count)
    else:
        print("\nGame Over!")
        print("The correct number was:", number)
    play = input("\nDo you want to play again? (yes/no): ")
    if play != "yes":
        print("\nThank you for playing!")
        break
```
